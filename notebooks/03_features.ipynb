# Features: what the models see

**What this answers:** Which inputs there are, how the new window and neighbourhood inputs are built, and how much is missing.

**What it reads:** The feature configuration, the modelling table, the enriched arrays and the daily weather series.

**What to expect:** A group table, an example of a window, a comparison of one input for fires and non-fires, and missing-value shares.

Load the reporting helpers.

```
from wildfire.reporting import features
from IPython.display import display
```

**Groups of inputs.** The 42 tabular inputs and the two new blocks.

```
display(features.group_table())
```

**One window feature explained.** The fire-weather index before one fire, with the 7-day and 30-day averages the model receives.

```
features.plot_window_example()
```

**Does an input separate fires from non-fires?** The fire-weather index for onsets and for matched days.

```
features.plot_feature_by_label('fwi')
```

**Missing values.** Rows with a missing value in any block are left out of every arm, so all arms use the same rows.

```
display(features.missing_table())
```

## What this shows

The enriched blocks add 209 weather-window columns and 12 neighbourhood columns to the 42 inputs. Fire days have a visibly higher fire-weather index than matched days.

## What it does not show

It does not show that the new inputs help the models; the results notebook answers that.
