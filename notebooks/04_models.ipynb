# Models: the ladder of arms

**What this answers:** Which models are compared, what each one reads and how the settings were chosen.

**What it reads:** The model configuration, the tuning trials and the chosen settings.

**What to expect:** An arm table, the tuning trials and the chosen settings.

Load the reporting helpers.

```
from wildfire.reporting import models
from IPython.display import display
```

**The ladder.** From a simple rate up to a network. The same rows, labels and folds are used for all.

```
display(models.ladder_table())
```

**Diagnostic arms.** They know only where a cell is. They show how much skill comes from place rather than from weather.

```
display(models.diagnostics_table())
```

**Tuning trials.** Twelve trials each for boosting and the network, judged on log loss over 2021 to 2024.

```
models.plot_tuning()
```

**Chosen settings.** The best trial of each model.

```
display(models.chosen_settings())
```

## What this shows

Boosting and the network had the same tuning budget, and the chosen settings were fixed before the final tables were made.

## What it does not show

It does not show which model is better; that is judged on validation years the tuning never used as a target.
